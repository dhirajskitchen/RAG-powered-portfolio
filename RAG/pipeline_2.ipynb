## Initialize Tools

```python
model = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")

embeddings = HuggingFaceEmbeddings(
    model_name="BAAI/bge-base-en-v1.5"
)


vector_store = Chroma(
    collection_name="portfolio",
    embedding_function=embeddings,
    persist_directory=DB_PATH
)
```

## Loading Docs

```python
loader = DirectoryLoader(
    path="../context_documents",
    glob="**/*.md",
    loader_cls=TextLoader
)

documents = loader.load()

for doc in documents:
    filepath = Path(doc.metadata["source"])

    doc.metadata.update({
        "filename": filepath.name,
        "category": filepath.stem,
        "filetype": "markdown"
    })

print(documents[0].metadata)

print(f"Total characters in doc 1: {len(documents[0].page_content)}")
```

## Chunking

```python
headers_to_split_on = [
    ("#", "section")
]

markdown_splitter = MarkdownHeaderTextSplitter(
    headers_to_split_on=headers_to_split_on
)

md_header_splits = []

for doc in documents:
    splits = markdown_splitter.split_text(doc.page_content)

    for split in splits:
        split.metadata.update(doc.metadata)
        md_header_splits.append(split)
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1200,
    chunk_overlap=200,
    add_start_index=True
)

final_chunks = text_splitter.split_documents(md_header_splits)

print(f"Total chunks: {len(final_chunks)}")
```

```python
for idx, chunk in enumerate(final_chunks):

    chunk.metadata["citation_id"] = f"C{idx:04d}"
    filename = chunk.metadata.get("filename", "").lower()

    if "personality" in filename:
        chunk.metadata["type"] = "personality"

    elif "skills" in filename or "projects" in filename:
        chunk.metadata["type"] = "technical"

    elif "faq" in filename:
        chunk.metadata["type"] = "recruiter"

    elif "opinions" in filename:
        chunk.metadata["type"] = "beliefs"

    else:
        chunk.metadata["type"] = "general"
```

## Reranker

### CrossEncoder Reranking using sentence_transformers

```python
from sentence_transformers import CrossEncoder
reranker = CrossEncoder(
    "cross-encoder/ms-marco-MiniLM-L-12-v2"
)

def rerank(query, docs, top_k=5):

    pairs = [
        (query, doc.page_content)
        for doc in docs
    ]

    scores = reranker.predict(pairs)

    ranked_docs = sorted(
        zip(docs, scores),
        key=lambda x: x[1],
        reverse=True
    )

    return [
        doc
        for doc, score in ranked_docs[:top_k]
    ]
```

## Hybrid retriever (BM25 + Vector Retriever)

```python
from langchain_community.retrievers import BM25Retriever

# BM25 retriever
bm25_retriever = BM25Retriever.from_documents(final_chunks)
bm25_retriever.k = 5

# Vector retriever
vector_retriever = vector_store.as_retriever(
    search_kwargs={"k": 5}
)
```

### Manual HybridRetriever

```python

def hybrid_search(query, k=5):

    bm25_docs = bm25_retriever.invoke(query)
    vector_docs = vector_retriever.invoke(query)

    scores = {}

    for rank, doc in enumerate(bm25_docs):
        scores[doc.page_content] = scores.get(doc.page_content, 0) + 1/(rank+60)

    for rank, doc in enumerate(vector_docs):
        scores[doc.page_content] = scores.get(doc.page_content, 0) + 1/(rank+60)

    doc_map = {
        doc.page_content: doc
        for doc in bm25_docs + vector_docs
    }

    ranked = sorted(
        scores.items(),
        key=lambda x: x[1],
        reverse=True
    )

    return [
        doc_map[content]
        for content, _ in ranked[:k]
    ]
```

```python


def retrieve(query):

    candidates = hybrid_search  (
        query,
        k=10
    )
    print(f"Number of candidates: {len(candidates)}")
        
    final_docs = rerank(
        query,
        candidates,
        top_k=5
    )

    return final_docs
```

## RAG Query Optimization

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate

model = ChatOllama(
    model="qwen3:4b",
    temperature=0
)

query_optimizer_prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system",
            """
You are a query optimization assistant for a RAG system.

The knowledge base contains information about a person named Dhiraj.

Your job is to rewrite the user's query so retrieval works better.

Rules:

1. Preserve the user's original intent.
2. Keep the query concise.
3. Expand abbreviations when useful.
4. Add relevant keywords only if they are strongly related.
5. Do not invent facts.
6. Do not answer the question.
7. Return ONLY the optimized search query.
8. If the query is already clear, return it unchanged.

Examples:

User: What are your hobbies?
Output: Dhiraj hobbies interests leisure activities

User: What tech stack do you use?
Output: Dhiraj technology stack programming languages frameworks tools

User: Where did you study?
Output: Dhiraj education academic background university college

User: Tell me about yourself
Output: Dhiraj profile background skills education projects experience
"""
        ),
        ("human", "{query}")
    ]
)

query_optimizer = query_optimizer_prompt | model


def optimize_query(query: str) -> str:
    response = query_optimizer.invoke(
        {"query": query}
    )

    optimized = response.content.strip()

    # Remove accidental thinking tags from Qwen
    optimized = optimized.replace("<think>", "")
    optimized = optimized.replace("</think>", "")
    optimized = optimized.strip()

    return optimized
```

## For Testing: Chatbot

```python
from langchain_ollama import ChatOllama
from langchain_core.messages import SystemMessage, HumanMessage
import re

llm = ChatOllama(
    model="qwen3:4b",
    temperature=0.1,
    num_ctx=8192
)


def generate_answer(query):

    docs = retrieve(query)

    if not docs:
        return "I do not have enough information."

    context = "\n\n".join(
        f"""
[CITATION: {doc.metadata.get('citation_id')}]

SOURCE:
{doc.metadata.get('filename')}

CONTENT:
{doc.page_content}
"""
        for doc in docs
    )

    valid_citations = {
        doc.metadata.get("citation_id")
        for doc in docs
        if doc.metadata.get("citation_id")
    }

    system_prompt = """
You are Dhiraj.

Answer naturally as Dhiraj would.

You MUST ONLY use information found in the provided context.

RULES:

1. Every factual sentence MUST end with at least one citation.
2. Never output a factual statement without a citation.
3. Use only citations provided in the context.
4. Do not invent citations.
5. Do not mention retrieval systems, vector databases, context documents,
   or the existence of provided context.
6. If the answer is not available in the context, reply exactly:

I do not have enough information.
"""

    user_prompt = f"""
CONTEXT:
{context}

QUESTION:
{query}

ANSWER:
"""

    response = llm.invoke(
        [
            SystemMessage(content=system_prompt),
            HumanMessage(content=user_prompt),
        ]
    )

    answer = response.content.strip()

    # Extract generated citations
    generated_citations = set(
        re.findall(r"\[(C\d+)\]", answer)
    )

    # Check for hallucinated citations
    invalid_citations = generated_citations - valid_citations

    if invalid_citations:
        print(
            f"WARNING: Invalid citations detected: "
            f"{sorted(invalid_citations)}"
        )

    # Check if answer contains citations
    if answer != "I do not have enough information.":
        if not generated_citations:
            print("WARNING: No citations generated")

    return answer


def chatbot():

    print("=" * 60)
    print("Portfolio RAG Chatbot")
    print("Type 'exit' to quit")
    print("=" * 60)

    while True:

        query = input("\nYou: ").strip()

        if query.lower() in ["exit", "quit"]:
            print("Goodbye!")
            break

        try:

            answer = generate_answer(query)

            print("\nBot:")
            print(answer)

        except Exception as e:

            print("\nError:")
            print(e)


chatbot()
```
